```python
# String Slicing Example

text = "Python Programming"

print("Original String:", text)

# ---------- Forward Slicing ----------
print("\nForward Slicing:")
print("text[0:6] =", text[0:6])      # Python
print("text[7:18] =", text[7:18])    # Programming
print("text[:6] =", text[:6])        # Python
print("text[7:] =", text[7:])        # Programming
print("text[::2] =", text[::2])      # Every 2nd character

# ---------- Backward Slicing ----------
print("\nBackward Slicing:")
print("text[::-1] =", text[::-1])    # Reverse string
print("text[17:6:-1] =", text[17:6:-1])
print("text[-1:-12:-1] =", text[-1:-12:-1])
print("text[::-2] =", text[::-2])    # Reverse with step 2
```
